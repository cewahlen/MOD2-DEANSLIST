```python
# Program to determine Dean's List and Honor Roll eligibility

while True:
    # Ask for student's last name
    last_name = input("Enter student's last name (or ZZZ to quit): ")

    # Exit the program if last name is ZZZ
    if last_name == "ZZZ":
        print("Program ended.")
        break

    # Ask for student's first name
    first_name = input("Enter student's first name: ")

    # Ask for student's GPA
    gpa = float(input("Enter student's GPA: "))

    # Check for Dean's List
    if gpa >= 3.5:
        print(f"{first_name} {last_name} has made the Dean's List.")

    # Check for Honor Roll
    if gpa >= 3.25:
        print(f"{first_name} {last_name} has made the Honor Roll.")

    print()  # Blank line between student records
```
